```python
# Import Necessary Libraries
import datetime as dt

# from pySpyEPI
from pySpyEPI.io.download import swarm_download as sdload
from pySpyEPI.io.load import swarm_load
from pySpyEPI.plotting import irr_plot
from pySpyEPI.detection import build
from pySpyEPI.barrel import barrel_utils as bar_uts
```

1. Download Swarm file

```python
# date for download
ymd = dt.datetime(2024, 6, 7)

# output file directory
out_dir = "~/SWARM"

# satellite 'A', 'B', or 'C'
satellite = 'B'

# use download function
# include swarm website URL, file level, baseline, instrument1, and instrument2
# this depends on what kind of file you are looking for. The example is for
# the Level 2 product IPIR dataset
# num_days is an integer indicating how many days you would like to download
# after ymd
sdload.download_and_unzip(
        ymd, satellite, out_dir,
        s_url="https://swarm-diss.eo.esa.int/?do=download&file=swarm%2FLevel",
        level='2daily', baseline='Latest_baselines',
        instrument1='IPD', instrument2='IRR',
        f_end='0302', T1='000000', T2='235959', num_days=0)
```

2. Load File

```python
# Establish the directory and time period
fdir = '~/SWARM'
satellite = 'B'
start_date = dt.datetime(2024, 6, 7, 0, 0)
end_date = start_date + dt.timedelta(days=1)

# You can load either EFI or IRR files based on which files you have
data = swarm_load.load_IRR(start_date, end_date, satellite, fdir=fdir)

# this outputs a dataframe
print(data.columns)
```

3. Section data based on magnetic latitude

```python
# auroral_bound is our upper limit for the midlatitude trough
auroral_bound = 70
data_aur = data[abs(data['mlat']) < auroral_bound]
```

```python
# use find_all_gaps to get mlat segments
# this looks for where ther is a jump in the index to make separating segments
# easier
sw_idx = data_aur.index.values
lat_gaps = bar_uts.find_all_gaps(sw_idx)
print(lat_gaps)
```

3. Use build.find_cand do find EPI candidates <blockquote>
build.find_cand is very flexible and has many optional inputs. <br>
The required inputs are the data, satellite, and satellite number <blockquote>
data : dataframe <br>
observation : satellite used for observation i.e. swarm, dmsp used for saving <br>
satellite number : specific satellite number or letter for Swarm 'A', 'B', or 'C'.  </blockquote>


 

```python
i = 11
data1 = data_aur[lat_gaps[i]: lat_gaps[i + 1]]
nflag_mask = data1['ne_flag'] <= 2

barrel_df, cand_df, peaks, properties, pass_id, trough_lats = build.find_cand(
    data1, 'swarm', 'B', trough=True, equator_bound=35, auroral_bound=70, set_lat=50,
    num_barrels=5, barrel_start=8, det_filt=5, peak_width=6, x_type='time',
    x_str='time', e_up_large=0.1, e_lo_large=0.05, e_up_small=0.1,
    e_lo_small=0.05, upper_weight=2, lower_weight=1, scale_n=True,
    exp_inc=True, ex_scale=2, filter_barrel=True, svg_window=5, svg_poly=2,
    freq=1,
    fft_test=True, nest_test=True, prom_test=True, percent_test=True,
    wdr_test=True, utest_mask=None,
    min_height=10**4, min_perc=10, wdr50_set=0.995, wdr20_set=0.95,
    eia_state='unknown', pass_id=None, prom_height=5000,
    time_str='time', ne_str='ne', glat_str='lat', glon_str='lon',
    mlat_str='mlat', mlon_str='mlon', alt_str='altitude', lshell_str='l_shell',
    lt_str='lt', nflag_mask=nflag_mask)
```

```python
# limit data by trough lats
data_eq = data1.copy()
data_eq = data_eq[(data_eq['mlat'] <= max(trough_lats)) & (data_eq['mlat'] >= min(trough_lats))]

# add time to barrel_df
barrel_df['time'] = data_eq['time'].values
```

```python
fig = irr_plot.ne_2barrel(data_eq, 'B', barrel_df, peaks, properties,
                          cand_df, EFI_data=None, hspace=0.4,
                          fig_size=None, leg_loc='center left', fs=14,
                          flag_mark=True, eia_state=None, eia_lats=None,
                          y_sup=0.96, x_sup=0.9)
```
